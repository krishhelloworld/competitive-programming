```python
def  numbers (first,*args):
    """this is a program to explain multiple arguement program"""
    if not args:
        return first
    max_rest = numbers(*args)
    return first if first > max_rest else max_rest
```

```python
print(numbers(1,3,5,7,7))
```

```python
import string,itertools
def convert(message,num=0,mode="encrypt"):
    check_is_upper = [ i for i,x  in enumerate (message) if x.isupper()]
    ##I WILL UPGRADGE IT LIKE A MESSAGE APP WHERE I WILL SHOW ENCRYPT MESSAGE AFTER EVERY passwords set BETWEEN USER
    encrypt= [x for x in itertools.islice(string.ascii_lowercase,n,26) ]
    add = [x for x in itertools.islice(string.ascii_lowercase,0,n) ]
    encrypt.extend(add)
    """print(encrypt)
    print(string.ascii_lowercase)"""
    space = " "
    if mode == "encrypt":
        e_message = [encrypt[ord(xy.lower())-97] if xy.isalpha()  else space if ord(xy)==10 else xy for xy in message  ] 
    else:
        e_message = [encrypt[ord(xy.lower())-(97+2*n)] if xy.isalpha()  else space if ord(xy)==10 else xy for xy in message  ] 
    for i in check_is_upper:
        if i or i==0:
            e_message[i]=e_message[i].upper()
    temp=["".join(g) for k,g in itertools.groupby(e_message,bool) if k]##GROUP THE NON EMPTY(FALSE BOOL VALUE) STRINGS
    result_e ="\033[33mencrypt~    "+ "".join(temp)
    result_d="\033[35mdecrypt~ "+"".join(temp)
    return result_e if mode == "encrypt" else result_d
```
